### Cross-phylum trait prediction from social-niche embeddings

Two things happen in this notebook, in this order:

1. **Honesty check** - leave-one-phylum-out: a random forest trained on Traitar
   labels is tested on BacDive labels of a held-out phylum. Nothing about the
   focal phylum is ever in the training set. This is the only source of the AUC
   numbers (paper Fig. 2C) and it is *not* used to make predictions.
2. **The trait table** - one random forest per trait, fitted on *all* Traitar
   labels, applied to every OTU in the vocabulary. OTUs that already carry a
   Traitar label keep it (`source = Traitar`); the rest get the inferred value
   (`source = SNE_predicted`). The per-phylum AUC from step 1 is attached to
   every row so a prediction is never shown without its own error bar.

There is no reason to build one model per phylum for step 2: phylum hold-out is
an evaluation scheme, not a deployment scheme. Withholding a phylum only throws
away labels; the AUC it produces is what tells you how far to trust the
all-labels model on that phylum.

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
```

```python
# --------------------------------------------------------------------------- #
# configuration
# --------------------------------------------------------------------------- #
FULL_EMB_DIR = "./"
TAXMAP       = "taxmap_slv_ssu_ref_nr_138.2.txt"
TRAITOR_CSV  = "trait_predcit.csv"
BACDIVE_CSV  = "bacDive.csv"
AUC_CSV      = "../data/traits_predict/auc_res.csv"          # step 1 output
TABLE_CSV    = "../data/traits_predict/trait_table.csv"      # step 2 output
```

```python
# Embeddings covering the full ID set. Only these are evaluated at 21w; the
# smaller data sizes exist for social_niche only.
FULL_EMBS = {
    "social_niche": "social_niche_embedding_100.txt"}

BIG4 = ["Bacillota", "Bacteroidota", "Actinomycetota", "Pseudomonadota"]
BIG3 = BIG4[:3]

# Traitar spells it "D-Sorbitol", BacDive "sorbitol"; harmonised to "Sorbitol".
METABOLIC = ["Lactose", "Salicin", "Glycerol", "Melibiose",
             "Maltose", "Sucrose", "Trehalose", "Sorbitol"]
```

```python
TRAITS = {
    "Oxygen_Preference": (BIG4, 4, 3, False),
    "Gram_Status":       (BIG3, 6, 2, False),
    "Motility":          (BIG3, 6, 2, False),
    "Spore_Formation":   (BIG4, 6, 2, False),
    **{m: (BIG3, 4, 2, True) for m in METABOLIC},
}

MODEL = RandomForestClassifier(n_estimators=1000, random_state=0, oob_score=True,
                               n_jobs=-1, class_weight="balanced")
```

```python
# --------------------------------------------------------------------------- #
# loading helpers
# --------------------------------------------------------------------------- #
def load_embedding(path):
    emb = pd.read_csv(path, header=None, sep=" ", low_memory=False, index_col=0)
    return emb.drop(index="<unk>", errors="ignore")


_full_cache = {}


def get_full_embedding(name):
    """Full-size embeddings do not depend on the repeat, so load each one once."""
    if name not in _full_cache:
        _full_cache[name] = load_embedding(f"{FULL_EMB_DIR}/{FULL_EMBS[name]}")
    return _full_cache[name]


def iter_embeddings(n):
    """Yield (embedding name, data-size key, embedding) for one repeat."""
    for t in DATASIZES:
        if t == "21w":
            for name in FULL_EMBS:
                yield name, t, get_full_embedding(name)
        else:
            yield "social_niche", t, load_embedding(SUBSET_TMPL.format(t=t, n=n))


def load_taxonomy():
    tax = pd.read_csv(TAXMAP, sep="\t", low_memory=False)
    # accession = "<col0>.<col1>.<col2>" (vectorised instead of a per-row loop)
    acc = (tax.iloc[:, 0].astype(str) + "." +
           tax.iloc[:, 1].astype(str) + "." +
           tax.iloc[:, 2].astype(str))
    ranks = tax["path"].str.split(";", expand=True).iloc[:, :7]
    ranks.columns = ["k", "p", "c", "o", "f", "g", "s"]
    ranks.index = acc.values
    return ranks


def combine_labels(df, mapping, exclusive=False):
    """Collapse indicator columns into one categorical column.

    mapping: {column: label}, applied in order (first match wins).
    exclusive=True -> NaN unless exactly one indicator is set.
    """
    out = pd.Series(np.nan, index=df.index, dtype=object)
    for col, label in mapping.items():
        out[out.isna() & (df[col] == 1)] = label
    if exclusive:
        out[df[list(mapping)].sum(axis=1) != 1] = np.nan
    return out


def load_traitor():
    tr = (pd.read_csv(TRAITOR_CSV, index_col=0).astype(int).replace(3, 1)
            .rename(columns={"D-Sorbitol": "Sorbitol"}))
    out = pd.DataFrame({
        "Oxygen_Preference": combine_labels(
            tr, {"Aerobe": "aerobic", "Facultative": "facultatively",
                 "Anaerobe": "anaerobic"}, exclusive=True),
        "Gram_Status": combine_labels(
            tr, {"Gram negative": "negative", "Gram positive": "positive"},
            exclusive=True),
        "Motility": tr["Motile"],
        "Spore_Formation": tr["Spore formation"],
    })
    return out.join(tr[METABOLIC])


def metabolic_columns():
    """BacDive column name -> metabolic trait name (acid production only)."""
    agg = pd.read_csv(AGG_BAC_CSV)
    agg["level_3"] = agg["level_3"].str.capitalize()
    agg = agg[agg["level_3"].isin(METABOLIC) &
              (agg["level_2"] == "builds_acid_from") & (agg["type"] == 1)]
    return dict(zip(agg["terms"], agg["level_3"]))


def load_bacdive():
    tr = (pd.read_csv(BACDIVE_CSV, low_memory=False)
            .drop_duplicates(subset="16s_ID")
            .set_index("16s_ID")
            .replace({"NA": np.nan, "": np.nan, "-": "no", "+": "yes",
                      "+;NA": np.nan, "mixed": np.nan, "variable": np.nan,
                      "no;yes": np.nan, "negative;positive": np.nan,
                      "negative;variable": np.nan}))
    out = pd.DataFrame({
        "Oxygen_Preference": combine_labels(
            tr, {"aerobe": "aerobic", "facultative.anaerobe": "facultatively",
                 "anaerobe": "anaerobic"}),
        "Gram_Status": tr["gram_stain"],
        "Motility": tr["motility"],
        "Spore_Formation": tr["spore_formation"],
    })
    out = out.join(tr[list(metabolic_columns())].rename(columns=metabolic_columns()))
    return out.replace({"yes": 1, "no": 0})


def align_bacdive(bacdive, fid):
    """Re-index BacDive (keyed by bare accession) onto embedding IDs."""
    acc = np.array([str(x).split(".")[0] for x in fid])
    keep = np.isin(acc, bacdive.index.values)
    out = bacdive.loc[acc[keep]]
    out.index = fid[keep]
    return out
```

## Step 1 - leave-one-phylum-out AUC

Train on Traitar, test on BacDive, holding out the focal phylum *and* every
phylum outside the focal set. Produces `data/auc_res.csv`.

```python
def score_split(emb, tax, traitor, bacdive, trait, phylum, focal, min_count, min_classes):
    """Train on Traitor, test on BacDive, holding out `phylum` + all non-focal phyla."""
    test_lab = bacdive[trait].dropna()
    tax_test = tax.loc[test_lab.index]
    held_out = set(p for p in tax_test["p"].unique() if p not in focal) | {phylum}

    test_id = tax_test.index[tax_test["p"].isin(held_out)]
    train_lab = traitor[trait].dropna()
    tax_train = tax.loc[train_lab.index]
    train_id = tax_train.index[~tax_train["p"].isin(held_out)]

    y_test = test_lab.loc[test_id].values
    classes, counts = np.unique(y_test, return_counts=True)
    if len(classes) < min_classes or counts.min() < min_count:
        return None

    MODEL.fit(emb.loc[train_id], train_lab.loc[train_id].values)
    proba = MODEL.predict_proba(emb.loc[test_id])
    if not set(classes).issubset(MODEL.classes_):
        return None                                   # test label unseen in training

    if len(MODEL.classes_) > 2:
        return roc_auc_score(y_test, proba, multi_class="ovr", average="macro",
                             labels=MODEL.classes_)
    positive = MODEL.classes_[1]
    return roc_auc_score((y_test == positive).astype(int), proba[:, 1])
```

```python
fid_ref = np.unique(np.concatenate(
    [load_embedding(REF_EMB_TMPL.format(n=n)).index.values for n in REPEATS]))

emb = load_embedding("social_niche_embedding_100.txt")
fid_ref = fid_ref[fid_ref != "<unk>"]

tax_all = load_taxonomy()
traitor_all = load_traitor()
bacdive_all = load_bacdive()

records = []
fid = np.intersect1d(fid_ref, emb.index.values)
emb = emb.loc[fid]

tax = tax_all.loc[fid]
traitor = traitor_all.loc[np.intersect1d(fid, traitor_all.index.values)]
bacdive = align_bacdive(bacdive_all, fid)

for trait, (focal, min_count, min_classes, full_only) in TRAITS.items():
    if full_only and t != "21w":
        continue
    for phylum in focal:
        auc = score_split(emb, tax, traitor, bacdive, trait, phylum,
                          focal, min_count, min_classes)
        if auc is not None:
            records.append({"auc": auc, "embedding": emb_name,
                            "traits_type": trait,
                            "group": f"times_{n}", "tax": phylum,
                            "datasize": DATASIZES[t]})

pd.DataFrame(records).to_csv(AUC_CSV, index=False)
```

## Step 2 - the trait table

One forest per trait, fitted on every Traitar label there is, applied to all
14k OTUs. Traitar-labelled OTUs stay in the table with their own label; only
`source` distinguishes them from the inferred ones.

`{trait}_prob` is the forest's probability *for the value in that row* - for a
Traitar row it is therefore in-sample and close to 1, which is why `source` and
not `prob` is what tells you whether a value is measured or guessed.
`{trait}_auc` is the step-1 AUC for that OTU's own phylum (mean over repeats),
falling back to the trait's mean AUC for phyla outside the focal set.

```python
def auc_lookup():
    """{trait: {phylum: LOPO AUC, "_mean": mean over phyla}} for social_niche@21w."""
    a = pd.read_csv(AUC_CSV)
    a = a[(a["embedding"] == "social_niche") & (a["datasize"] == DATASIZES["21w"])]
    out = {}
    for trait, g in a.groupby("traits_type"):
        out[trait] = {**g.groupby("tax")["auc"].mean().to_dict(),
                      "_mean": g["auc"].mean()}
    return out
```

```python
emb = get_full_embedding("social_niche")
fid = emb.index.values

tax = load_taxonomy().reindex(fid)
traitor = load_traitor().astype(object).reindex(fid)   # keep 0/1 as ints, not 0.0/1.0
aucs = auc_lookup()

table = pd.DataFrame({"otu_id": fid, "phylum": tax["p"].values,
                      "genus": tax["g"].values})

for trait in TRAITS:
    known = traitor[trait]
    y = known.dropna()
    MODEL.fit(emb.loc[y.index], y.values)
    proba = MODEL.predict_proba(emb)
    col = {c: i for i, c in enumerate(MODEL.classes_)}

    value = np.where(known.notna(), known, MODEL.classes_[proba.argmax(1)])
    table[trait] = value
    table[f"{trait}_prob"] = [proba[i, col[v]] if v in col else np.nan
                              for i, v in enumerate(value)]
    table[f"{trait}_source"] = np.where(known.notna(), "Traitar", "SNE_predicted")
    a = aucs.get(trait, {"_mean": np.nan})
    table[f"{trait}_auc"] = [a.get(p, a["_mean"]) for p in table["phylum"]]

table.to_csv(TABLE_CSV, index=False)
table.head()
```

```python
# sanity: every OTU covered, Traitar rows reproduce their own label,
# and the labelled/inferred split is what we expect
assert len(table) == len(fid) and table["otu_id"].is_unique
for trait in TRAITS:
    src = table[f"{trait}_source"]
    lab = traitor[trait].dropna()
    assert (src == "Traitar").sum() == len(lab)
    assert (table.loc[src.values == "Traitar", trait].values == lab.values).all()

table[[f"{t}_source" for t in TRAITS]].apply(pd.Series.value_counts).T
```
